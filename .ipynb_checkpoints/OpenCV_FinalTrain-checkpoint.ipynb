```python
import h5py
import numpy as np
import os
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
```

```python
import seaborn as sb
import matplotlib.pyplot as plt
```

```python
X = np.array(h5py.File(os.path.join('OpenCV_Generated_File', 'Features_Harlick_Hue.h5'), 'r')['features'])
y = np.array(h5py.File(os.path.join('OpenCV_Generated_File', 'Folders_name.h5'), 'r')['targets'])
```

```python
np.shape(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=100)
```

```python
rfc = RandomForestClassifier(n_estimators=400)
rfc.fit(X_train, y_train)
```

```python
y_pred = rfc.predict(X_test)
```

```python
print(classification_report(y_pred=y_pred, y_true=y_test))
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn_model = KNeighborsClassifier(n_neighbors=3)
knn_model.fit(X_train, y_train)
```

```python
y_pred_knn = knn_model.predict(X_test)
```

```python
print(classification_report(y_pred=y_pred_knn, y_true=y_test))
```

```python
from sklearn import metrics
```

```python
print(metrics.confusion_matrix(y_test, knn_model.predict(X_test)))
```

```python
from sklearn.tree import DecisionTreeClassifier# tree is libraray and DecisionTreeClassifier is a class 
```

```python
dtree = DecisionTreeClassifier()# make new class or from constractur
dtree.fit(X_train, y_train)
```

```python
y_pred_dtree=dtree.predict(X_test)
```

```python
print(classification_report(y_pred=y_pred_dtree, y_true=y_test))
```

```python
print(metrics.confusion_matrix(y_test, y_pred_dtree))
```

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.svm import SVC
```

```python
ada_clf = AdaBoostClassifier(base_estimator=SVC(probability=True, kernel='rbf'), n_estimators=10)
ada_clf.fit(X_train, y_train)
```

```python
y_pred_ada=ada_clf.predict(X_test)
```

```python
print(classification_report(y_pred=y_pred_ada, y_true=y_test))
```

```python
confusion_matrix=metrics.confusion_matrix(y_test, y_pred_ada)
```

```python
cm = confusion_matrix
sb.heatmap(cm, annot = True, fmt = 'g')
plt.xlabel("Predicted")
plt.ylabel("Inputs")
plt.title("Confusion Matrix")
plt.show()
```

```python
SVM_model=SVC(kernel='rbf')
```

```python
SVM_model.fit(X_train, y_train)
```

```python
y_pred_SVM=SVM_model.predict(X_test)
```

```python
print(classification_report(y_pred=y_pred_SVM, y_true=y_test))
```

```python
confusion_matrix_SVM=metrics.confusion_matrix(y_test, y_pred_SVM)
```

```python
cm = confusion_matrix_SVM
sb.heatmap(cm, annot = True, fmt = 'g')
plt.xlabel("Predicted")
plt.ylabel("Inputs")
plt.title("Confusion Matrix")
plt.show()
```
