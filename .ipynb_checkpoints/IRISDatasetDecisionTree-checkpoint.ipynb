```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df.head()
```

```python
len(df)
```

```python
y = iris.target
count_0=0
for i in range(1, len(y)+1):
    if y[i-1]==1:
        count_0+=1
print(count_0)
```

```python
from sklearn.tree import DecisionTreeClassifier# tree is libraray and DecisionTreeClassifier is a class 
```

```python
dtree = DecisionTreeClassifier()# make new class or from constractur
# dtree.fit(df, y)
```

```python
# dtree.predict([
#     [6.1, 2.8, 1.8, 1.9],
#     [8.1, 2.1, 2.5, 2.9]
# ])
```

```python
X_train=df[:120]
y_train=y[:120]
X_test=df[120:]
y_test=y[120:]
```

```python
dtree.fit(X_train, y_train)
```

```python
y_predicted=dtree.predict(X_test)
```

```python
count_true=0
count_false=0
for i in range(len(y_test)):
    if(y_predicted[i]==y_test[i]):
        count_true+=1
    else:
        count_false+=1
print("count_true", count_true, "count_false", count_false)
```

```python
from sklearn import metrics
```

```python
print(metrics.classification_report(y_test, y_predicted))
```

```python
# from sklearn.externals.six import StringIO
from six import StringIO
from IPython.display import Image
from sklearn.tree import export_graphviz
import pydotplus# for installing pydotplus first of all install this package then install graphvis
```

```python
dot_data = StringIO()
export_graphviz(dtree, out_file=dot_data, filled=True, rounded=True, special_characters=True)
graph = pydotplus.graph_from_dot_data(dot_data.getvalue())
Image(graph.create_png())
```
