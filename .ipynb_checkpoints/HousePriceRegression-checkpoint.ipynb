```python
import xgboost as xgb
import numpy as np
```

```python
X = data.iloc[:, 3:].values#train and Test data; features
y = data['price'].values# That's target house price
```

```python
np.shape(y)
```

```python
np.shape(X)
```

```python
data_matrix = xgb.DMatrix(data=X, label=y)# data is features and label is target in XGB function
```

```python
data_matrix
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=100)
```

```python
xg_reg = xgb.XGBRegressor(colsample_bytree=0.8, learning_rate=0.1, max_depth=6, n_estimators=1000, verbosity=3)
```

```python
xg_reg.fit(X_train, y_train)
```

```python
y_pred = xg_reg.predict(X_test)# xg_reg is my xgboost model for this data
```

```python
from sklearn import metrics
```

```python
metrics.mean_absolute_error(y_test, y_pred)
```

```python
y_pred_XGB = xg_reg.predict(X_test)
data3 = pd.DataFrame({"Y_test": y_test , "Y_pred" : y_pred_XGB})
data3.head(10)
```

```python
plt.figure(figsize=(12,8))
plt.plot(data3[:50])
plt.legend([ 'Actual' ,'Predicted'])
```

```python
metrics.r2_score(y_test, y_pred)# That's useful for regression question and if close 1, so the algorithm is powerfull
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import RandomizedSearchCV
```

```python
params = {
    'colsample_bytree': [0,2,4],
    'learning_rate': [0, .1,.2 ],
    'max_depth': [3,4,6],
    'n_estimators': [300,800]  
}
gsh = GridSearchCV(xgb.XGBRegressor(), param_grid=params, scoring='r2', cv=3, n_jobs=-1, verbose=3)
```

```python
gsh.fit(X, y)
```
