```python
import matplotlib.pyplot as plt
import cv2
import numpy as np
import mahotas
```

```python
cv2
```

```python
img = cv2.imread('train_data/Bluebell/11.jpg')
```

```python
np.shape(img)
```

```python
print(img)
```

```python
hist = cv2.calcHist([img],[0],None,[256],[0,256])
```

```python
plt.plot(hist, color='b')
plt.title('Image Histogram For Blue Channel GFG')
plt.show()
```

```python
color = ('b','g','r')
for i,col in enumerate(color):
    histr = cv2.calcHist([img],[i],None,[256],[0,256])
    plt.plot(histr,color = col)
    plt.xlim([0,256])
plt.show()
```

```python
img2 = cv2.imread('train_data/daisy/2.jpg')
```

```python
color = ('b','g','r')
for i,col in enumerate(color):
    histr = cv2.calcHist([img2],[i],None,[256],[0,256])
    plt.plot(histr,color = col)
    plt.xlim([0,256])
plt.show()
```
