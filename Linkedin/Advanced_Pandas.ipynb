```python
# Time Series
```

```python
datarange=pd.period_range('1/1/2020', freq='30d', periods=4)
```

```python
date_df=pd.DataFrame(data=datarange, columns=['sample_date'])
date_df
```

```python
date_df['date_difference']=date_df['sample_date'].diff(periods=1)
date_df
```

```python
date_df['first of month']=date_df['sample_date'].values.astype('datetime64[M]')
date_df
```

```python
date_df.dtypes
```

```python
date_df['Time_Stamped_date']=date_df['sample_date'].dt.to_timestamp()
date_df
```

```python
date_df.dtypes
```

```python
date_df['Time_Stamped_date']-date_df['first of month']
```
