```python
filename, _ = request.urlretrieve('https://www.robots.ox.ac.uk/~vgg/data/flowers/17/17flowers.tgz', 'flowers.tgz')
```

```python
print(os.stat(filename))
```

```python
tar = tarfile.open(filename)
tar.extractall(path='train_data_matin_images')
tar.close()
```

```python
image_paths = sorted(glob.glob('train_data_matin_images/*.jpg'))
```

```python
image_paths
```

```python
folders = [
    'Buttercup', 'ColtsFoot', 'Daffodil', 'Daisy', 'Dandelion', 'Fritillary', 'Iris', 'Pansy', 'Sunflower', 'Windflower',
    'Snowdrop', 'LilyValley', 'Bluebell', 'Crocus', 'Tigerlily', 'Tulip', 'Cowslip'
]
l = 0
i=0
j=80
for k in range(1, 18):
    os.makedirs('train_data_matin_images/' + folders[l])
    current_path = 'train_data_matin_images/' + folders[l] + '/'
    for index, image_path in enumerate(image_paths[i:j], start=1):
        original = image_path # train_data_matin_images\\image_0052.jpg
        image_path = image_path.split('/')#[train_data_matin_images,image_0052.jpg ]
        image_file_name = str(index) + '.jpg'
        os.rename(original, current_path + image_file_name)
    l += 1 # new folder
    i += 80 # new package start
    j += 80 # new package end
```
