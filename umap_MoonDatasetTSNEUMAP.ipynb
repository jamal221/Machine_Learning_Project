```python
X, y = make_moons(n_samples=1000, noise=0.01, random_state=258)
set(y)
```

```python
tsne = TSNE(n_components=1, perplexity=50 ,random_state=123)# Create model
tsne_result = tsne.fit_transform(X)# Fit and transform model
tsne_result# This data is modeled
```

```python
tsne_result.shape
```

```python
plt.scatter(tsne_result[:, 0], range(0, len(tsne_result)), c=y)
```

```python
plt.scatter(tsne_result[:, 0], np.ones(len(tsne_result)), c=y)
```
