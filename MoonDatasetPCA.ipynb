```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA, KernelPCA
from sklearn.datasets import make_moons
from matplotlib import pyplot as plt
```

```python
X, y = make_moons(n_samples=1000, noise=0.01)
# X.shape
y.shape
```

```python
set(y)
# by this code we can check value of y, it looks like uinque() function
```

```python
# y[300:800]=1
# This code can be applied for check color, in scatter we have to 
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y)
# X[:, 0] is X axis and X[:, 1] is y axis, the first argumant belongs to x and the second place is for y  
```

```python
pca = PCA(n_components=1, random_state=123)
pca.fit(X)
pca_result = pca.transform(X)
pca_result
```

```python
plt.scatter(pca_result[:, 0], range(0, len(pca_result)), c=y)
```

```python
plt.scatter(pca_result[:, 0], np.ones(len(pca_result)), c=y)
```

```python
kpca = KernelPCA(n_components=1, kernel='poly', gamma=10, random_state=123)
kpca.fit(X)
kpca_result = kpca.transform(X)
kpca_result
```

```python
plt.scatter(kpca_result[:, 0], range(0, len(kpca_result)), c=y)
```
